Distribution plots help us understand how numerical (quantitative) data is spread across a dataset. They reveal patterns such as central tendency, variability, skewness, and outliers.
distplot/histplot
* jointplot
* pairplot
* rugplot

```python
import pandas as pd

import matplotlib.pyplot as plt 
dataset = pd.read_csv("../tip.csv")
print(dataset.head(2))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.subplot(1,2,1)
sns.histplotplot(dataset["total_bill"], bins=20, kde=True)
plt.subplot(1,2,2)
sns.histplotplot(dataset["tip"], bins=20, kde=True)
```

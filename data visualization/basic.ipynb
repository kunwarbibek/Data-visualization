```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
x=np.linspace(0,5,11)

y=x**2
print(x,y)


plt.plot(x,y)
plt.title("  Baisc Lab")
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

A subplot divides a figure into a grid and allows you to plot multiple graphs in different sections of the same figure.
nrows → number of rows in the grid
ncols → number of columns in the grid
index → position of the current plot (starts from 1)

```python
from numpy import poly
plt.subplot(2, 2, 1)
plt.subplot(2, 2, 2)
plt.plot(y,x)
plt.subplot(2, 2, 3)
plt.plot(x, y)
plt.subplot(2, 2, 4)
```
